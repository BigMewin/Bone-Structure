```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LassoCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_percentage_error
from sklearn.model_selection import train_test_split

# Function to calculate accuracy as 1 - MAPE
def calculate_accuracy(y_true, y_pred):
    mape = mean_absolute_percentage_error(y_true, y_pred)
    return 1 - mape

# Function to calculate Concordance Correlation Coefficient (CCC) with outlier exclusion
def concordance_correlation_coefficient_with_outliers(y_true, y_pred, std_threshold=2):
    residuals = np.abs(y_true - y_pred)
    threshold = np.mean(residuals) + std_threshold * np.std(residuals)
    inliers = residuals < threshold

    mean_true = np.mean(y_true[inliers])
    mean_pred = np.mean(y_pred[inliers])
    var_true = np.var(y_true[inliers])
    var_pred = np.var(y_pred[inliers])
    covariance = np.mean((y_true[inliers] - mean_true) * (y_pred[inliers] - mean_pred))
    ccc = (2 * covariance) / (var_true + var_pred + (mean_true - mean_pred) ** 2)

    return ccc, inliers

# Function to plot BvTv vs Tb.Sp with a regression line and mark outliers
def plot_bvtv_vs_tbsp(df_before, df_after, outliers, df_name):
    plt.figure(figsize=(12, 6))
    
    # Plot before outlier removal
    plt.subplot(1, 2, 1)
    plt.scatter(df_before['BvTv'], df_before['Tb.Sp'], alpha=0.6, label='Inliers')
    plt.scatter(df_before['BvTv'][outliers], df_before['Tb.Sp'][outliers], color='red', label='Outliers')
    plt.title(f'BvTv vs Tb.Sp Before Outlier Removal ({df_name})')
    plt.xlabel('BvTv')
    plt.ylabel('Tb.Sp')
    slope_before, intercept_before = np.polyfit(df_before['BvTv'], df_before['Tb.Sp'], 1)
    plt.plot(df_before['BvTv'], slope_before * df_before['BvTv'] + intercept_before, color='red', linestyle='--', label='Regression Line')
    plt.legend()
    plt.grid(True)
    
    # Plot after outlier removal
    plt.subplot(1, 2, 2)
    plt.scatter(df_after['BvTv'], df_after['Tb.Sp'], alpha=0.6, color='orange')
    plt.title(f'BvTv vs Tb.Sp After Outlier Removal ({df_name})')
    plt.xlabel('BvTv')
    plt.ylabel('Tb.Sp')
    slope_after, intercept_after = np.polyfit(df_after['BvTv'], df_after['Tb.Sp'], 1)
    plt.plot(df_after['BvTv'], slope_after * df_after['BvTv'] + intercept_after, color='red', linestyle='--', label='Regression Line')
    plt.legend()
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()

# Prepare empty variables for reduced DataFrames
merged_df_reduced_1 = None
merged_df_reduced_11 = None
merged_df_reduced_2 = None
merged_df_reduced_22 = None
merged_df_reduced_3 = None
merged_df_reduced_33 = None

# Range of percentiles to test
percentile_range = np.arange(50, 96, 5)

# Iterate through each DataFrame for outlier identification and processing
for df_name in ['merged_df1', 'merged_df11', 'merged_df2', 'merged_df22', 'merged_df3', 'merged_df33']:
    df = globals()[df_name]  # Get the DataFrame by name

    # Step 1: Outlier detection in BvTv vs Tb.Sp using standard deviation threshold
    X_bvtv = df[['BvTv']].values
    y_tbsp = df['Tb.Sp'].values

    # Linear regression to get predictions for Tb.Sp
    slope, intercept = np.polyfit(X_bvtv.flatten(), y_tbsp, 1)
    y_pred_tbsp = slope * X_bvtv.flatten() + intercept
    residuals_tbsp = y_tbsp - y_pred_tbsp

    # Define outliers based on a standard deviation threshold
    threshold_tbsp = np.mean(residuals_tbsp) + 2 * np.std(residuals_tbsp)
    outliers_tbsp = np.abs(residuals_tbsp) > threshold_tbsp

    # Remove BvTv vs Tb.Sp outliers from the dataset
    df_no_outliers = df[~df.index.isin(df.index[outliers_tbsp])]
    
    # Store reduced DataFrames in variables
    if df_name == 'merged_df1':
        merged_df_reduced_1 = df_no_outliers
    elif df_name == 'merged_df11':
        merged_df_reduced_11 = df_no_outliers
    elif df_name == 'merged_df2':
        merged_df_reduced_2 = df_no_outliers
    elif df_name == 'merged_df22':
        merged_df_reduced_22 = df_no_outliers
    elif df_name == 'merged_df3':
        merged_df_reduced_3 = df_no_outliers
    elif df_name == 'merged_df33':
        merged_df_reduced_33 = df_no_outliers

    # Step 2: Perform LassoCV predictions and evaluate metrics before and after outlier removal
    prediction_results_before = []
    prediction_results_after = []

    # Features and labels for metrics
    X_all = df.iloc[:, 6:].values
    y_metrics = df.iloc[:, 2:6].values
    X_scaled_all = StandardScaler().fit_transform(X_all)

    X_no_outliers = df_no_outliers.iloc[:, 6:].values
    y_metrics_no_outliers = df_no_outliers.iloc[:, 2:6].values
    X_scaled_no_outliers = StandardScaler().fit_transform(X_no_outliers)

    # Loop through each metric and perform LassoCV with CCC calculation
    for i, metric_name in enumerate(df.columns[2:6]):
        y = y_metrics[:, i]
        y_no_outliers = y_metrics_no_outliers[:, i]

        # Split data into training and test sets
        X_train, X_test, y_train, y_test = train_test_split(X_scaled_all, y, test_size=0.2, random_state=42)
        X_train_no_outliers, X_test_no_outliers, y_train_no_outliers, y_test_no_outliers = train_test_split(
            X_scaled_no_outliers, y_no_outliers, test_size=0.2, random_state=42)

        # LassoCV prediction before outlier removal
        lasso_cv = LassoCV(cv=5, max_iter=10000000)
        lasso_cv.fit(X_train, y_train)
        y_pred = lasso_cv.predict(X_test)
        mse = mean_squared_error(y_test, y_pred)
        accuracy = calculate_accuracy(y_test, y_pred)
        ccc, _ = concordance_correlation_coefficient_with_outliers(y_test, y_pred)

        prediction_results_before.append({
            'Dataset': df_name,
            'Metric': metric_name,
            'MSE': mse,
            'Accuracy': accuracy,
            'CCC': ccc
        })

        # LassoCV prediction after outlier removal
        lasso_cv_no_outliers = LassoCV(cv=5, max_iter=10000000)
        lasso_cv_no_outliers.fit(X_train_no_outliers, y_train_no_outliers)
        y_pred_no_outliers = lasso_cv_no_outliers.predict(X_test_no_outliers)
        mse_no_outliers = mean_squared_error(y_test_no_outliers, y_pred_no_outliers)
        accuracy_no_outliers = calculate_accuracy(y_test_no_outliers, y_pred_no_outliers)
        ccc_no_outliers, _ = concordance_correlation_coefficient_with_outliers(y_test_no_outliers, y_pred_no_outliers)

        prediction_results_after.append({
            'Dataset': df_name,
            'Metric': metric_name,
            'MSE': mse_no_outliers,
            'Accuracy': accuracy_no_outliers,
            'CCC': ccc_no_outliers
        })

    # Convert prediction results to DataFrames and display
    print(f"\nResults Before Outlier Removal for {df_name}:")
    print(pd.DataFrame(prediction_results_before))
    print(f"\nResults After Outlier Removal for {df_name}:")
    print(pd.DataFrame(prediction_results_after))

    # Plot BvTv vs Tb.Sp before and after outlier removal
    plot_bvtv_vs_tbsp(df, df_no_outliers, outliers_tbsp, df_name)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LassoCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# Function to calculate accuracy as 1 - MAPE
def calculate_accuracy(y_true, y_pred):
    mape = np.mean(np.abs((y_true - y_pred) / y_true))  # MAPE calculation
    return 1 - mape

# Function to calculate Concordance Correlation Coefficient (CCC) with outlier exclusion
def concordance_correlation_coefficient_with_outliers_excluded(y_true, y_pred, std_threshold=2):
    # Calculate residuals and define threshold for outliers
    residuals = np.abs(y_true - y_pred)
    threshold = np.mean(residuals) + std_threshold * np.std(residuals)
    
    # Identify non-outliers
    inliers = residuals < threshold
    
    # Calculate CCC for inliers only
    mean_true = np.mean(y_true[inliers])
    mean_pred = np.mean(y_pred[inliers])
    var_true = np.var(y_true[inliers])
    var_pred = np.var(y_pred[inliers])
    covariance = np.mean((y_true[inliers] - mean_true) * (y_pred[inliers] - mean_pred))
    ccc = (2 * covariance) / (var_true + var_pred + (mean_true - mean_pred) ** 2)
    
    return ccc, inliers

# Function to plot true vs predicted values for each metric in 4 subplots, marking outliers
def plot_true_vs_pred(y_true, y_pred, title_suffix, dataset_type, outliers):
    metrics = ['BvTv', 'Tb.Th', 'Tb.Sp', 'Tb.N']
    plt.figure(figsize=(14, 10))
    
    for i, metric in enumerate(metrics):
        plt.subplot(2, 2, i + 1)
        plt.scatter(y_true[:, i], y_pred[:, i], alpha=0.6, color='blue', label='Inliers')
        
        # Mark outliers in red
        outlier_indices = ~outliers[i]
        plt.scatter(y_true[:, i][outlier_indices], y_pred[:, i][outlier_indices], color='red', label='Outliers')
        
        plt.xlabel(f'True {metric}')
        plt.ylabel(f'Predicted {metric}')
        plt.title(f'{metric} - True vs Predicted ({dataset_type})')
        plt.plot([min(y_true[:, i]), max(y_true[:, i])],
                 [min(y_true[:, i]), max(y_true[:, i])], color='red', linestyle='--')
        plt.legend()
        plt.grid(True)
    
    plt.suptitle(f'True vs Predicted Values - {title_suffix} ({dataset_type})')
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])  # Adjust layout to make room for the main title
    plt.show()

# Dictionary for bone data
bones = {
    "bone_1": [merged_df_reduced_1, merged_df_reduced_11],
    "bone_2": [merged_df_reduced_2, merged_df_reduced_22],
    "bone_3": [merged_df_reduced_3]
}

# Loop through each train-test configuration
train_test_combinations = [
    #(["bone_1", "bone_2"], "bone_3"),
    #(["bone_1", "bone_3"], "bone_2"),
   # (["bone_2", "bone_3"], "bone_1"),
    (["bone_1"], "bone_2")
]

# Standardize the features
scaler = StandardScaler()

# Initialize a list to store results for the summary DataFrame
summary_results = []

# Loop over each train-test combination
for train_bones, test_bone in train_test_combinations:
    print(f"\nTraining on bones: {train_bones}, Testing on bone: {test_bone}\n")
    
    # Prepare train and test sets
    train_dfs = [df for bone in train_bones for df in bones[bone]]
    train_data = pd.concat(train_dfs).reset_index(drop=True)

    # Get the reduced and full datasets for the test bone
    test_dfs_reduced = bones[test_bone]
    test_dfs_full = [globals()[f'merged_df{test_bone[-1]}'], globals()[f'merged_df{test_bone[-1]}{test_bone[-1]}']]
    
    for test_type, test_dfs in [("Reduced", test_dfs_reduced), ("Full", test_dfs_full)]:
        print(f"Testing on bone {test_bone} - {test_type} Dataset\n")

        test_data = pd.concat(test_dfs).reset_index(drop=True)

        # Define X and y for both train and test sets
        X_train = train_data.iloc[:, 6:].values
        y_train = train_data[['BvTv', 'Tb.Th', 'Tb.Sp', 'Tb.N']].values

        X_test = test_data.iloc[:, 6:].values
        y_test = test_data[['BvTv', 'Tb.Th', 'Tb.Sp', 'Tb.N']].values

        # Standardize the features
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)

        # Initialize lists to store results
        metrics_results = []
        outliers_per_metric = []

        # Apply LassoCV and calculate metrics
        y_pred_test = np.zeros_like(y_test)  # Initialize array to store predictions for plotting
        for i, metric_name in enumerate(['BvTv', 'Tb.Th', 'Tb.Sp', 'Tb.N']):
            # LassoCV model
            lasso_cv = LassoCV(cv=5, max_iter=10000000)
            lasso_cv.fit(X_train_scaled, y_train[:, i])
            y_pred_test[:, i] = lasso_cv.predict(X_test_scaled)

            # Calculate metrics
            mse = mean_squared_error(y_test[:, i], y_pred_test[:, i])
            accuracy = calculate_accuracy(y_test[:, i], y_pred_test[:, i])
            ccc, inliers = concordance_correlation_coefficient_with_outliers_excluded(y_test[:, i], y_pred_test[:, i])

            metrics_results.append({
                'Metric': metric_name,
                'MSE': mse,
                'Accuracy': accuracy,
                'CCC': ccc
            })
            outliers_per_metric.append(inliers)

            # Append to summary results for the final chart
            summary_results.append({
                'Train Bones': ', '.join(train_bones),
                'Test Bone': test_bone,
                'Test Dataset Type': test_type,
                'Metric': metric_name,
                'MSE': mse,
                'Accuracy': accuracy,
                'CCC': ccc
            })

        # Convert metrics to DataFrame and print results
        metrics_df = pd.DataFrame(metrics_results)
        print(f"\nResults for bone {test_bone} ({test_type} Dataset):")
        print(metrics_df)

        # Plot true vs predicted values for all metrics, marking outliers
        plot_true_vs_pred(y_test, y_pred_test, f"Train on {', '.join(train_bones)}; Test on {test_bone}", test_type, outliers_per_metric)

# Convert summary results to a DataFrame and display it
summary_results_df = pd.DataFrame(summary_results)
print("\nFinal Summary of All Combinations:")
print(summary_results_df)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
from sklearn.model_selection import train_test_split

# Function to calculate accuracy as 1 - MAPE
def calculate_accuracy(y_true, y_pred):
    mape = mean_absolute_percentage_error(y_true, y_pred)
    return 1 - mape

# Function to calculate Concordance Correlation Coefficient (CCC) with outlier exclusion
def concordance_correlation_coefficient_with_outliers(y_true, y_pred, std_threshold=2):
    residuals = np.abs(y_true - y_pred)
    threshold = np.mean(residuals) + std_threshold * np.std(residuals)
    inliers = residuals < threshold

    mean_true = np.mean(y_true[inliers])
    mean_pred = np.mean(y_pred[inliers])
    var_true = np.var(y_true[inliers])
    var_pred = np.var(y_pred[inliers])
    covariance = np.mean((y_true[inliers] - mean_true) * (y_pred[inliers] - mean_pred))
    ccc = (2 * covariance) / (var_true + var_pred + (mean_true - mean_pred) ** 2)

    return ccc, inliers

# Prepare empty variables for reduced DataFrames
merged_df_reduced_1 = None
merged_df_reduced_11 = None
merged_df_reduced_2 = None
merged_df_reduced_22 = None
merged_df_reduced_3 = None
merged_df_reduced_33 = None

# Iterate through each DataFrame for outlier identification and processing
for df_name in ['merged_df1', 'merged_df11', 'merged_df2', 'merged_df22', 'merged_df3', 'merged_df33']:
    df = globals()[df_name]  # Get the DataFrame by name

    # Step 1: Outlier detection in BvTv vs Tb.Sp
    X_bvtv = df[['BvTv']].values
    y_tbsp = df['Tb.Sp'].values

    slope, intercept = np.polyfit(X_bvtv.flatten(), y_tbsp, 1)
    y_pred_tbsp = slope * X_bvtv.flatten() + intercept
    residuals_tbsp = y_tbsp - y_pred_tbsp

    threshold_tbsp = np.mean(residuals_tbsp) + 2 * np.std(residuals_tbsp)
    outliers_tbsp = np.abs(residuals_tbsp) > threshold_tbsp

    df_no_outliers = df[~df.index.isin(df.index[outliers_tbsp])]

    # Store reduced DataFrames in variables (NOT DELETED!)
    if df_name == 'merged_df1':
        merged_df_reduced_1 = df_no_outliers
    elif df_name == 'merged_df11':
        merged_df_reduced_11 = df_no_outliers
    elif df_name == 'merged_df2':
        merged_df_reduced_2 = df_no_outliers
    elif df_name == 'merged_df22':
        merged_df_reduced_22 = df_no_outliers
    elif df_name == 'merged_df3':
        merged_df_reduced_3 = df_no_outliers
    elif df_name == 'merged_df33':
        merged_df_reduced_33 = df_no_outliers

    # Step 2: Perform Linear Regression and evaluate metrics before and after outlier removal
    prediction_results_before = []
    prediction_results_after = []

    # Define feature mapping
    feature_mapping = {
        'BvTv': 'original_glrlm_RunLengthNonUniformity',
        'Tb.Th': 'original_glcm_ClusterTendency',
        'Tb.Sp': 'original_glrlm_GrayLevelVariance',
        'Tb.N': 'original_glcm_ClusterShade'
    }

    # Loop through each metric and perform Linear Regression with CCC calculation
    for metric_name in ['BvTv', 'Tb.Th', 'Tb.Sp', 'Tb.N']:
        predictor = feature_mapping[metric_name]

        # Use the appropriate predictor for each metric
        X_all = df[[predictor]].values
        y_metrics = df[[metric_name]].values
        X_scaled_all = StandardScaler().fit_transform(X_all)

        X_no_outliers = df_no_outliers[[predictor]].values
        y_metrics_no_outliers = df_no_outliers[[metric_name]].values
        X_scaled_no_outliers = StandardScaler().fit_transform(X_no_outliers)

        # Split data into training and test sets
        X_train, X_test, y_train, y_test = train_test_split(X_scaled_all, y_metrics, test_size=0.2, random_state=42)
        X_train_no_outliers, X_test_no_outliers, y_train_no_outliers, y_test_no_outliers = train_test_split(
            X_scaled_no_outliers, y_metrics_no_outliers, test_size=0.2, random_state=42)

        # Linear Regression before outlier removal
        model = LinearRegression()
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        mse = mean_squared_error(y_test, y_pred)
        accuracy = calculate_accuracy(y_test, y_pred)
        ccc, _ = concordance_correlation_coefficient_with_outliers(y_test, y_pred)

        prediction_results_before.append({
            'Dataset': df_name,
            'Metric': metric_name,
            'MSE': mse,
            'Accuracy': accuracy,
            'CCC': ccc
        })

        # Linear Regression after outlier removal
        model_no_outliers = LinearRegression()
        model_no_outliers.fit(X_train_no_outliers, y_train_no_outliers)
        y_pred_no_outliers = model_no_outliers.predict(X_test_no_outliers)
        mse_no_outliers = mean_squared_error(y_test_no_outliers, y_pred_no_outliers)
        accuracy_no_outliers = calculate_accuracy(y_test_no_outliers, y_pred_no_outliers)
        ccc_no_outliers, _ = concordance_correlation_coefficient_with_outliers(y_test_no_outliers, y_pred_no_outliers)

        prediction_results_after.append({
            'Dataset': df_name,
            'Metric': metric_name,
            'MSE': mse_no_outliers,
            'Accuracy': accuracy_no_outliers,
            'CCC': ccc_no_outliers
        })

    # Convert prediction results to DataFrames and display
    print(f"\nResults Before Outlier Removal for {df_name}:")
    print(pd.DataFrame(prediction_results_before))
    print(f"\nResults After Outlier Removal for {df_name}:")
    print(pd.DataFrame(prediction_results_after))
```
